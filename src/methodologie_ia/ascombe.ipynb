```python
import seaborn as sns
import matplotlib.pyplot as plt
```

# Anscombe

```python
df = sns.load_dataset("anscombe")

print(df.nunique())
print(df.info())
df.head()
```

```python
fig, ax = plt.subplots(2,2 , sharex=True, sharey=True)
gs = fig.add_gridspec(2,2)

ax1 = fig.add_subplot(gs[0,0])
ax2 = fig.add_subplot(gs[0,1], sharey=ax1)
ax3 = fig.add_subplot(gs[1,:], sharex=ax1)

sns.histplot(df, x='x', ax=ax1)
sns.histplot(df, x='y', ax=ax2)
sns.scatterplot(df, x='x', y='y', ax=ax3, hue='dataset')

plt.tight_layout()
plt.show()
```

```python
sns.relplot(df, x='x', y='y', col='dataset')

plt.show()
```

```python
df.groupby('dataset').describe()
```

* Verifier que les donnée continue n'appatienne pas a un groups: Sous une variable ce cache souvent plusieur signaux
* Les statistique sont parfois trompeuse -> Francis Anscombe (1973) a voulu démontrer que le fait de construire des graphiques était une activité utiles et importantes!
